```python
from google.colab import userdata
from huggingface_hub import whoami, InferenceClient

token = userdata.get("HF_TOKEN")
print("user:", whoami(token=token)["name"])
```

```python
import os
from google.colab import userdata
from smolagents import GoogleSearchTool

os.environ["SERPER_API_KEY"] = userdata.get("SERPER_API_KEY")
search = GoogleSearchTool(provider="serper")
print(search("best party music recommendations"))
```

```python
from smolagents import WebSearchTool

#print(WebSearchTool()("best party music recommendations"))   # default engine
print(WebSearchTool(engine="bing")("best party music recommendations"))
```

```python
from smolagents import WebSearchTool
print(WebSearchTool(engine="duckduckgo")("best party music recommendations"))
```

```python
from smolagents import ToolCallingAgent, WebSearchTool, OpenAIServerModel
from google.colab import userdata

model = OpenAIServerModel(
    model_id="gpt-4o-mini",
    api_key=userdata.get("OPENAI_API_KEY"),
)
agent = ToolCallingAgent(tools=[WebSearchTool()], model=model)
agent.run("Search for the best music recommendations for a party at Wayne's mansion.")
```
